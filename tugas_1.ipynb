Pelanggaran Hukum: Di Indonesia, tindakan ini melanggar Pasal 27A dan Pasal 28 ayat (1) UU ITE (UU No. 1 Tahun 2024) terkait pencemaran nama baik dan penyebaran berita bohong yang merugikan, serta ketentuan perlindungan data pribadi dalam UU No. 27 Tahun 2022 (UU PDP). Kasus nyata yang sering terjadi adalah penipuan berbasis voice cloning untuk meminta uang kepada keluarga korban.
